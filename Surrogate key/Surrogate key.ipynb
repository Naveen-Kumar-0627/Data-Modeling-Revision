## Why we dont relay on natural key 
<p> if we do join using customer_id we got duplicate rows.
</p>
<p>one order was happend but we got two rows
</p>
<p> Because natural key customer_id is not unique,and it will not any more</p>


```python
%sql
select order_id,city,amount from dim_customer as c join fact as f on c.customer_id=f.customer_id;
```

## The Surrogate key
<p> surrogate key is unique so if do join using this ,we can solve ambiguity,redundancy</p> 


```python
%sql
select order_id,city,amount from dim_customer as c join fact as f on c.customer_sk=f.customer_sk;
```
